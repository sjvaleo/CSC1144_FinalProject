```python
import pandas as pd

# Load u.data
ratings = pd.read_csv("u.data", sep="\t", names=["user_id", "movie_id", "rating", "timestamp"])
ratings.head()
```

```python
# Create user-item matrix: users as rows, movies as columns, ratings as values
user_item_matrix = ratings.pivot_table(index='user_id', columns='movie_id', values='rating')

# Fill missing values with (not rated)
user_item_matrix.fillna(0, inplace=True)

# print the matrix
user_item_matrix.head()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Compute cosine similarity between users
user_similarity = cosine_similarity(user_item_matrix)

# Put it into a DataFrame
user_similarity_df = pd.DataFrame(user_similarity,
                                  index=user_item_matrix.index,
                                  columns=user_item_matrix.index)

# print  the similarity matrix
user_similarity_df.head()
```

```python
def get_user_recommendations(user_id, top_n_similar=5, num_recommendations=5):
    # Step 1: Find top N most similar users
    similar_users = user_similarity_df[user_id].sort_values(ascending=False).iloc[1:top_n_similar+1]

    # Step 2: Get their ratings
    similar_users_ratings = user_item_matrix.loc[similar_users.index]

    # Step 3: Weighted sum of ratings based on similarity scores
    weighted_ratings = similar_users_ratings.T.dot(similar_users)

    # Step 4: Normalize by total similarity score
    recommendation_scores = weighted_ratings / similar_users.sum()

    # Step 5: Remove movies already seen by the user
    user_seen = user_item_matrix.loc[user_id][user_item_matrix.loc[user_id] > 0].index
    recommendation_scores = recommendation_scores.drop(user_seen, errors='ignore')

    # Step 6: Return top movie IDs and scores
    return recommendation_scores.sort_values(ascending=False).head(num_recommendations)
```

```python
# Get top recommendations for user 1
recs = get_user_recommendations(user_id=1)

# Load the movie titles from u.item
movies = pd.read_csv("u.item", sep="|", encoding="latin-1", header=None,
                     names=["movie_id", "title", "release_date", "video_release_date", "IMDb_URL",
                            "unknown", "Action", "Adventure", "Animation", "Children's", "Comedy", "Crime",
                            "Documentary", "Drama", "Fantasy", "Film-Noir", "Horror", "Musical", "Mystery",
                            "Romance", "Sci-Fi", "Thriller", "War", "Western"])

# Show recommended movie titles
print(f"\nTop recommended movies for User 1:\n")
for movie_id, score in recs.items():
    title = movies[movies["movie_id"] == movie_id]["title"].values[0]
    print(f"{title} — Score: {score:.2f}")
```

- Evaluation Metrics:

```python
from sklearn.model_selection import train_test_split

# Split the data: 80% train, 20% test
train_data, test_data = train_test_split(ratings, test_size=0.2, random_state=42)

# Build training matrix
train_matrix = train_data.pivot_table(index='user_id', columns='movie_id', values='rating').fillna(0)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Create user-item matrix from training set
train_user_item = train_matrix.copy()

# Compute cosine similarity between users based on training data
train_user_similarity = cosine_similarity(train_user_item)

# Turn it into a DataFrame
train_user_similarity_df = pd.DataFrame(train_user_similarity,
                                        index=train_user_item.index,
                                        columns=train_user_item.index)
```

```python
from math import sqrt
from tqdm import tqdm

def predict_rating(user_id, movie_id, top_n=5):
    if user_id not in train_user_similarity_df.index or movie_id not in train_user_item.columns:
        return np.nan  # unknown user or movie

    # Get similarity scores for the user
    similarities = train_user_similarity_df[user_id]

    # Get ratings from other users for this movie
    movie_ratings = train_user_item[movie_id]

    # Combine similarity and rating
    valid_users = movie_ratings[movie_ratings > 0].index
    if len(valid_users) == 0:
        return np.nan

    similarities = similarities[valid_users]
    ratings = movie_ratings[valid_users]

    # Get top N similar users
    top_similar_users = similarities.sort_values(ascending=False).head(top_n)
    top_ratings = ratings[top_similar_users.index]

    if top_similar_users.sum() == 0:
        return np.nan

    # Weighted average
    predicted_rating = np.dot(top_ratings, top_similar_users) / top_similar_users.sum()
    return predicted_rating

# Predict ratings for test set
actual_ratings = []
predicted_ratings = []

print("Predicting ratings... (this may take a minute)")
for _, row in tqdm(test_data.iterrows(), total=len(test_data)):
    actual = row["rating"]
    predicted = predict_rating(row["user_id"], row["movie_id"])
    if not np.isnan(predicted):
        actual_ratings.append(actual)
        predicted_ratings.append(predicted)
```

```python
from sklearn.metrics import mean_squared_error

# Calculate RMSE
rmse = sqrt(mean_squared_error(actual_ratings, predicted_ratings))
print(f"RMSE on test set: {rmse:.4f}")
```

```python
from sklearn.neighbors import NearestNeighbors

# Fit KNN on user-item matrix
knn = NearestNeighbors(metric='cosine', algorithm='brute')
knn.fit(user_item_matrix)

# Find top 5 similar users for a given user
user_vector = user_item_matrix.loc[1].values.reshape(1, -1)
distances, indices = knn.kneighbors(user_vector, n_neighbors=6)

# Show similar users
print("Top similar users (excluding self):")
for i in range(1, len(indices[0])):
    similar_user_id = user_item_matrix.index[indices[0][i]]
    similarity_score = 1 - distances[0][i]
    print(f"User {similar_user_id} — Similarity: {similarity_score:.2f}")
```

- Help to Find Movie ID from Movie Name

```python
# Helper function to search for a movie title and return the movie ID(s)
def find_movie_id(title_query, case_insensitive=True):
    if case_insensitive:
        matched = movies[movies["title"].str.contains(title_query, case=False, na=False)]
    else:
        matched = movies[movies["title"].str.contains(title_query, na=False)]

    if matched.empty:
        print(f"No match found for: '{title_query}'")
    else:
        print(f"Matches for '{title_query}':\n")
        for _, row in matched.iterrows():
            print(f"{row['title']} — ID: {row['movie_id']}")
```
